```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
for i, document in enumerate(document):
    print(f"Document {i + 1}: {document}")
```

```python
document = [
    "Machine Learning enables computers to learn from data.",
    "ML is a branch of Artificial Intelligence.",
    "Machine Learning helps computers identify patterns.",
    "ML algorithms can make predictions using data.",
    "Machine Learning is widely used in modern technology."
]

vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(document)

print(tfidf_matrix)
```

```python
print(tfidf_matrix.shape)
```

```python
print(tfidf_matrix)
```

```python
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=["D1","D2","D3","D4", "D5"]
)

tfidf_df.round(2)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)
```

```python
similarity_matrix
```

```python
similarity_df = pd.DataFrame(
    similarity_matrix,
    columns=["D1", "D2", "D3", "D4", "D5"],
    index=["D1", "D2", "D3", "D4", "D5"]
)
similarity_df.round(3)
```

```python
highest_similarity = -1
most_similar_pair = None

for i in range(len(document)):
    for j in range(i + 1, len(document)):

        similarity = similarity_matrix[i][j]

        if similarity > highest_similarity:
            highest_similarity = similarity
            most_similar_pair = (i, j)

print(
    "Most similar documents: "
    f"D{most_similar_pair[0] + 1} and D{most_similar_pair[1] + 1}"
)

print(f"Similarity: {highest_similarity:.3f}")
```
