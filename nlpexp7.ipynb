```python
arr()
import nltk
from nltk.chunk import ne_chunk
nltk.download('maxent_ne_chunker_tab')
named_entities = ne_chunk(tagged)
print(named_entities)
```

```python
arr()
#Display named entities
for subtree in named_entities:
  entity = ""
  label = ""
  if isinstance(subtree, nltk.Tree):
    # This block handles actual named entities like (PERSON Barack Obama)
    entity_words = [word for word, tag in subtree.leaves()]
    entity = " ".join(entity_words) # Join with space for readability
    label = subtree.label()
  else:
    # This block handles individual word/tag tuples
    entity = subtree[0]
    label = subtree[1]

  print(f"Entity : {entity} , Label : {label}")
```
